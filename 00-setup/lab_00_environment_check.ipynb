```python
import numpy as np

area = np.array([50, 70, 90, 110, 130])
price = np.array([100, 140, 180, 220, 260])

print("Areas:", area)
print("Prices:", price)
```

```python
import matplotlib.pyplot as plt

plt.scatter(area, price, label="Training data")
plt.xlabel("Area (m²)")
plt.ylabel("Price (thousands)")
plt.title("House Prices vs Area")
plt.legend()
plt.grid(True)
plt.show()
```
